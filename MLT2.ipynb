```python

# here to implement simple liner regression will perform some steps
#step 1 : Importing required libraries
#Step 2 : Read the data
#Step 3 : Model selection and fitting
#Steo 4 : Predicting the value using linear equation y=mx+c
#Step 5: Visualizing the data
```

**Step 1 : Impoting Libraries**

```python
#Step 1 : Importing libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

**Step 2 : Read the data**

```python
data ={"hours": [1,2,3,4,5],
       "Marks": [50,60,80,90,100]}
df=pd.DataFrame(data)
print(df)
```

```python
x= df[["hours"]].values
y= df[["Marks"]].values
```


**Model Fitting**

equation for linear regression is y=mx+c what is slope it is y/x means,change in deoendent variable when there is unit change in independent variable.

**What is intercept** =even in the absence of x there will be some value i.e even if the student is not studing he/she will be getting some marks that is basically intercept according to this example.

so basically here we are finding change in   marks when these is unit change in no of hours studied



```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x,y)
```

```python
slope=model.coef_[0]
intercept=model.intercept_
```

```python
#"f-string" (formatted string literal) in python
#".2f: This is a format specifier used within the curly braces of an f-string
print(f"Slope(m): {slope}")
print(f"Intercept (b): {intercept}")
print(f"Equation of the regressionline: y={slope[0]:.2f} x + {intercept[0]:.2f}")
```

```python
df['Predicted Marks']=model.predict(x)
```

```python
#here keep in mind the data which we have provided to the model is also linear
df.head()
```

**Visualizing**

```python
plt.scatter(df['hours'],df['Marks'],color='blue',label='Actual Marks')
```

**plotting the equation**

```python
plt.plot(df['hours'],df['Predicted Marks'],color='red',label='Regression Line')
plt.xlabel('Hours of Study')
plt.ylabel('Marks')
plt.title('Linear Regression:Hours vd marks')
plt.legend()
plt.show()
```
